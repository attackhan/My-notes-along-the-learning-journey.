Python'da en sık kullanılan veri tipleri :
1- Tamsayılar (Integers)
2-Ondalıklı Sayılar ( Floats )
3- Karakter Dizileri ( Strings )

Python'a 5 sayısının tırnak içerisindeki "5" olarak girersek, bu ifadeyi bir karakter yani string olarak algılayacaktır. Eğer bu ifadeyle matematiksel bir işlem yapmayı denerseniz başarısız oluruz. Bu durumda "5" ifadesini,5 sayısına çevirmemiz lazım. 

Bu nedenle str() , int() ve float() fonksiyonları bulunmaktadır.

str() fonksiyonu parametre olarak girilen değeri karakter dizisine çevirir.

int() fonksiyonu,parametre olarak girilen değeri tamsayıya çevirir.

float() fonksiyonu,parametre olarak girilen değeri ondalık sayıya çevirir.

Bazen matematiksel işlemleri gerçekleştirirken çok fazla sayıda değeri aynı işleme tabi tutak zorunda kalabiliriz. Örneğin 2,3,4,7,11,19 sayılarının her birinin,bir diğeri ile toplamının hesaplandığı bir programa ihtiyaç olduğu varsayılsın, Bunun için 2 + 3,2 + 4,2 + 7 + ... , 11 + 19 şeklinde her bir işelmi teker teker yazmak hem geliştiriciyi yoracak hem de kod kalabalğına sebep olacaktır. 

Diğer probleme benzer şekilde,bir matematiksel işlemin aynı programda farklı sayılarla tekrar edilmesi gerekebilir. Örneğin, yazdığımız programda 2* farklı satırda, toplama işlemi gerçekleşiyor olsun. Programda kullanılacak sayıların değişmesi durumunda, tek tek ilgili 20 satırı aramak ve sayıları değiştirmek yine oldukça yorucu bir süreç oluşmasına sebep olabilir.
