Karşılaştırma operatörleri :
Operatörler           Görevi
 ==                   İki değer birbirine eşitse True,değilse False değer döner.
 !=                   İki değer birbirine eşit değilse True,diğer durumda True döner.
 >                    Soldaki değer sağdaki değerden büyüksa True,değilse False döner
 <                    Soldaki değer sağdaki değerden küçükse True,değilse False döner
 >=                   Soldaki değer sağdaki değerden büyükse veya sağdaki değere eşitse True değilse                            False döner
 <=                   Soldaki değer sağdaki değerden küçükse veya sağdaki değere eşitse True,değilse                            False döner

```python
"Mehmet" == "Mehmet"
```

```python
"Mehmet" == "Murat"
```

```python
"Mehmet" != "Murat"
```

```python
"Araba" < "kadayıf"
```

MANTIKSAL BAĞLAÇLAR 
Mantıksal bağlaçlar daha çok karşılaştırma işlemini kontrol ettiğimiz zamanlarda kullanılır.

and Operatörü
bu mantıksal bağlaç,bütün karşılaştırma işlemlerinin sonucunun True olmasına bakar. Bağlanan karşılaştırma işlemlerinin hepsinin kendi içinde sonucu True ise genel sonuç True, diğer durumlarda ise sonuç False çıkar. Kullanımı bu şekildedir.

```python
1 < 2 and "Murat" == "Murat"
```

```python
2 > 3 and "Murat" == "Murat"
```

```python
2 == 2 and 3.14 < 2.54 and "Elma" != "Armut"
```

```python
2 != 3 and 3.14 > 2.54 and "Elma" != "Armut"
```

### İşlemlerin birinin bile sonucu False ise genel işlemin sonucu False çıkmaktadır

### or Operatörü
Bu mantıksal bağlaç,bütün karşılaştırma işlemlerinin sonuçlarından en az birinin True olmasına bakar. Bağlanan karşılaştırma işlemlerinin en az birinin True olmasında genel sonuç True, diğer durumlarda ise sonuç False çıkar.

```python
1 < 2 or "Murat" != "Murat"
```

```python
1 == 2 or "Murat" == "Murat"
```

```python
1 > 3 or "Atakan" != "Atakan"
```

### not Operatörü
not operatörü aslında mantıksal bir bağlaç değildir. Bu operatör sadece bir mantıksal değeri veya karşılaştırma işleminin tam tersi sonuca çevririr. Yani,not operatörü True olan bir sonucu False, False olan bir snucu True sonucuna çevirir. 

```python
not 2 == 2
```

```python
not 2 != 2
```

```python
not "Python" == "Python"
```
