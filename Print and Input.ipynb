The Print function is a function that allows the value/values entered as parameters to be written to the screen. We've defined a value to more of a variable, and what does that turn into? We use it most for this purpose.We use it to see the results we want.

```python
print("hi everyone")
```

```python
print ("Hi","Everyone")
```

No direct value needs to be entered into the Print function. It can be assigned to a variable and entered as a parameter for this variable.

```python
x = "everyone"
print("hi",x)
```

```python
y = "hi" 
print(y,x)
```

The printing function automatically separates the values it receives with spaces. To separate values using a different delimiter, such as a comma or hyphen, this must be specified in the printing function. 

```python
print("Hi", "Everyone", sep = ",")
```

```python
print("Hi", "Everyone")
```

```python
print("Hi","Everyone", sep = "-")
```

Print fonksiyonu içerisinde başka fonksiyonlar da çalıştırılabilir. Örneğin, karakter diziler(stingler) için özelleştirilmiş format fonksiyonu, içerisinde en yaygın kullanılan fonksiyonlardan biridir. 

Format fonksiyonu bir string içerisinde dinamik olarak değişmesi gereken bir değerin kolayca değişebilmesi için kullanılır. Bunun sürekli değişiklik gösterecek değer yerine boş bir küme parantezi "{}" yerleştirilir. Karakter dizisinin birtirme tırnağı kapatıldıktan sonra '.format()' ifadesi yerleştirilir ve parantez içerisinde o an istenen değer girilir. Örnek ; A ve B isminde iki para birimi olsun. A para biriminin değeri sabitken B para biriminin değerinin sürekli güncellenmesi gereksin. Bu durumda yazılması gereken ifade şu şekilde olur :

"1 birim A'nın değeri {} birim B'ye eşittir".format("3")

Yukarıdaki ifadenin çıktısı şu şekilde olacaktır. 
1 birim A'nın değeri 3 birim B'ye eşittir.

Kısa bir süre sonra B para biriminin değeri değişsin. Bu durumda format fonksiyonu kullanıldığında yapılması gereken tek şey girilen değeri değiştirmek olacaktır. Böyle bir durumda güncellenen ifade şu olur : 
"1 birim A'nın değeri {} birim B'ye eşittir".format("5)

Yukarıdaki İfadenin çıktısı şu şekilde olacaktır.
1 birim A'nın değeri 5 birim B'ye eşittir.
