# Metodlar

### Metod nedir ?
Şimdiye kadar Pythonda kullanabildiğimiz bir çok veri tipi gördük ve bazı veritipleri üzerinde bu veritiplerinin metodlarını kullandık. Aslında bu veritiplerinin oluşturulan her bir değişken Pythonda obje ( object ) olarak düşünülür ve Python geliştiricileri bu objelere birçok metod tanımlamıştır.

Metodlar bir obje üzerinde belli işlemleri gerçekleştiren objelere özgü fonksiyonlardır ve objelerin üzerinde metodlar şu şekilde kullanılır:

obje.herhangi_bir_metod(değerler(opsiyonel))


```python
# Örnek bir liste objesi oluşturduğumuz zaman bu objeniin üzerinde belli metodları uygulayabiliriz.

liste = [1,2,3,4,5,6,7]
liste.insert(1,"Murat")
print(liste)
```

```python
a = [1,2,3,4,5,6,7,8]
a.insert(4,"Atakan")
print(a)
```

```python
a.append("Mehmet")
print(a)
```

```python
a.pop()
a.pop()
```

```python
print(a)
```

# Fonksiyonlar

## Parametreler ve Argümanlar

selam fonksiyonunun içine herhangi bir değer gönderemiyorduk ve fonksiyonumuz hep aynı işi yapıyordu. Ancak çoğu zaman fonksiyonlarımız içine gönderdiğimiz değerlere farklı işlemler yaparlar. 

```python
def selam(isim):   # isim değişkenimiz burada parametre olmaktadır.
    print("Merhaba:",isim)
```

```python
selam("Atakan") # "Atakan" değeri burada argüman olmaktadır
```

```python
selam("Mehmet")
```

```python
def selamla(isim):
    print("İsminiz:",isim)
```

```python
selamla("Atakan")
```

```python
def selamlama(isim):
    print("İsminiz:",isim)
```

```python
selamlama("Atakan")
```

```python
def toplama(a,b,c):
    print("Toplamları:",a+b+c)
    
```

```python
toplama(5,12,13)
```

```python
def faktoriyel(sayı):
    faktoriyel = 1
    if sayı == 0 or sayı == 1:
        print("Faktöriyel",faktoriyel)
    else:
        while(sayı >= 1):
            faktoriyel *= sayı
            sayı -= 1
        print("Faktöriyel:",faktoriyel)
```

```python
faktoriyel(1)
```

```python
faktoriyel(0)
```

```python
faktoriyel(5)
```
