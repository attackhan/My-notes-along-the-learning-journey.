SÖZLÜKLER 

Sözlükler veya ingilizce ismiyle dictionaryler aynı gerçek hayattaki sözlükler gibi davranan bir veri tipidir.Bu veritipi, şimdiye kadar gördüğümüz tüm veritiplerinden yapısı gereği farklıdır. Sözlüğün içindeki her bir eleman indeks ile değil, anahtar ( key),değer (value) olarak tutulur. Bu anlamda gerçek hayattaki sözlüklere oldukça benzerler. Örneğin,elimizde bir ingilizce - Türkçe sözlük alıp freedom kelmisini (key ya da anahtar) aradığımız zaman karşılık değer özgürlük ( değer ya da value ) olarak karşımıza çıkar. Sözlükleri de bu şekilde düşünebiliriz.

SÖZLÜK OLUŞTURMAK 

```python
#süslü parantez ve iki nokta ile anahtar değerlerimizi yerleştirelim

sözlük1 = {"Sıfır":0,"Bir":1,"İki":2,"Üç":3}
```

```python
sözlük1
```

```python
type(sözlük1)
```

```python
#İçine hiçbir eleman koymazsak sözlük yine de oluşturabiliriz ya da dict() fonksiyonu ile de oluşturabiliriz örnek :
sözlük1 = {}
sözlük2 = dict()
```

```python
sözlük1
sözlük2
```

```python
sözlük1 = {"Sıfır":0,"Bir":1,"İki":2,"Üç":3}
```

```python
#misal sözlük1 1 anahtarına ulaşmak istiyelim. bunun için basitçe sözlük1["bir"] anahtarını vermemiz yeterli
sözlük1["Bir"]
```

```python
#demetlerden farklı olarak sözlüklere eleman ekleyebiliriz.
sözlük1["Beş"] = 5
```

```python
sözlük1
```

```python
a = {"bir": [1,2,3,4], "iki": [[1,2],[3,4],[5,6]],"üç": 15}
```

```python
a["iki"]
```

```python
a["iki"][1][1]
```

```python
#Sözcükteki eleman değerini yani karşılık gelen anahtar kelimeyi değiştirebiliriz.Ve değer ekleyip çıkarabiliriz.
sözlük1["Beş"] = 10
```

```python
sözlük1["Beş"] 
```

```python
sözlük1["Beş"] + 1
```

TEMEL SÖZLÜK METODLARI

```python
yeni_sözlük = {"bir":1,"iki":2,"üç":3}
```

```python
#values() metodu sözlüğün değerlerini(value) bir liste olarak döner. 
yeni_sözlük.values()
```

```python
#keys metodu sözlüğün anahtarlarını(key) bir liste olarak döner.
yeni_sözlük.keys()
```

```python
#items() metodu sözlüğün anahtar ve değerlerini bir liste içinde demet olarak döner.
yeni_sözlük.items()
```
