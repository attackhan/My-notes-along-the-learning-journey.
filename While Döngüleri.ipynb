# While Döngüleri

    while döngüleri belli bir koşul sağlandığı sürece bloğundaki işlemleri gerçekleştirmeye devam eder. while döngülerinin sona ermesi için koşul durumunun bir süre sonra False olması gereklidir.Yapısı şu şekildedir :
                while (koşul):
                        işlem1
                        işlem2
                        işlem3
                          //
                          //
                          
             

```python
### Döngüde i değerlerini ekrana yazdırma
i = 0
while (i<10):
    print("i'nin değeri:",i)
    i+=1  #koşulun bir süre sonra false olması için gerekli 
```

```python
i = 0 
while(i < 255):
    print("i'nin değeri:",i)
    i += 1 
```

```python
i = 0
while(i < 20):
    print("i'nin değeri:",i)
    i += 2
```

```python
a = 0
while(a<40):
    print("Python öğreniyorum....")
    a += 1
```

```python
liste = [1,2,3,4,5]
for i in liste:
    print(i)
```

```python
index = 0 
while(index < len(liste)):
    print("İndex:",index,"Liste Elemanı:",liste[index])
    index += 1
```

# range() Fonksiyonu
    Pythondaki bu hazır fonksiyon bizim verdiğimiz değerlere göre range isimli bir yapı oluşturur ve bu yapı listelere oldukça benzer. Bu yapı başlangış,bitiş ve opsiyonel olarak artırma değeri alarak listelere benzeyen bir sayı dizisi oluşturur. 

```python
range(0,20) # 0'dan 20'a kadar (dahil değil) sayı dizisi oluşturur.
```

```python
print(*range(0,20)) # Yazdırmak için başına "*" koymamız gerekiyor.
```

```python
liste = list(range(0,20)) # list fonksiyonuyle listeye dönüştürülebilir.
```

```python
liste
```

```python
print(*range(5,10))
```

```python
print(*range(30)) # Başlangıç değeri vermediğimizde 0'dan başlar
```

```python
print(*range(1,100,2)) #3.kod atlama sayısını verir
```

```python
print(*range(5,100,5))
```

print(*range(20,0)) # range fonksiyonu içine hiçbir artırma eylemi verilmediği zaman yukarı doğru gitme eğilimindedir.Bu yüzden geri gitmek için bir azaltma değeri vermek gerekir

```python
print(*range(20,0,-1))
```

```python
#range() fonksiyonunu for döngülerinde kullanma
liste = [0,1,2,3,4,5]
for i in liste:
    print(i)
```

```python
for i in range(1,101):
    print(i)
```

#### range() fonksiyonu ile for döngüleri ilerletilebilir

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,10):
    print("*" * i)
```

```python
for i in range(1,100):
    print("*" * i)
```
