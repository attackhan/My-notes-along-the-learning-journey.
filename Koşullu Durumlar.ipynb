### Python Programlarının çalışma mantığı
Python programları çalışmaya başladığı zaman kodlarımızın yukarıdan başlayarak teker teker çalıştırılır ve çalıştırılacak kod kalmayınca programımız soona erer. Şöyle bir örneği düşünelim.

```python
a = 2 
b = 3 
c = 4
print(a+b+c)
```

#### Yukarıdaki basit kodda program teker teker her bir satırı ve ifadeyi çalıştırır ve çalıştırılacak kod kalmayınca program sona erer. Ancak Python'da her program bu kadar basit olmayabilir. Çoğu zaman Python programlarımız belirli bloklardan oluşur ve bloklar her zaman çalışmak zorunda olmaz. Peki bu bloklar nasıl tanımlanır ? Python'da bir blok tanımlama işlemi Girintiler sayesinde olmaktadır. Örnek açısından, Pythonda bloklaş şu şekilde olabilir :

```python
a == 2 # Blok 1'e ait kod
if (a  == 2 ) :
    print(a) # Blok 2'ye ait kod
print("merhaba") # Blok 1 e ait kod
```

```python
### Girintiler Tab tuşuyla yapılır
```

 #### if'in bulunduğu satırdan sonraki print işlemini bir tab kadar girintili yazdık. Burada gördüğümüz gibi girintiler(Tab) Pythonda bir blok oluşturmak için kullanılıyor ve her bloğunun çalıştırılması gerekmiyor. Altaaki örnekteki gibi :

```python
a == 2 # Blok 1'e ait kod 
if (a == 3 ) :
    print(a) # Blok 2'ye ait kod
    print("Atakan") # Blok 2'ye ait kod
print("Merhaba") # Blok 1'e ait kod
```

# KOŞULLU DURUMLAR 

##### Koşullu durumlar aslında günlük yaşamda sürekli karşılaştığımız durumlardır. Örneğin havanın yağmurlu olma koşuluna göre şemsiyemizi alırız veya uykumuzun gelme koşuluna göre uyuruz. Aslında programlamada birçok koşullu durumla karşılaşırız. Örneğin : belli koşullara göre belli işlemleri yaparız, belli koşullara göre yapmayız. İşte bunlar koşullu durumların temelini oluşturur.  

## İf Bloğu
if bloğu programımızın içinde herhangi bir yerde belli bir koşulu kontrol edecek isek kullanılan bloklardır. Yazımı şu şekildedir: 
        İf (koşul)
        # İf bloğu - koşul sağlanınca ( True ) çalışır. Bu hizadaki her işlem if bloğuna ait
        # if bloğu - Girintiyle oluşturulur.
        yapılacak işlemler
        
İf bloğu eğer koşul sağlanmışsa anlam taşır. Eğer if kalıbındaki koşul sağlanırsa ( True ) if bloğu çalıştırılır. Koşul sağlanmazsa ( False ) if bloğu çalıştırılmaz. 



```python
#18 yaş kontorlü

yaş = int(input("Yaşınızı giriniz:"))
if (yaş < 18):
    print("Bu mekana giremezsiniz.")
```

## Else Bloğu :
else blokları if koşulu sağlanmadığı zaman ( False ) çalışan bloklardır. Kullanımı şu şekildedir.
        Else : 
        # Else bloğu - yukarısındaki herhangi bir if bloğu  ( veya ileride göreceğimiz elif bloğu ) çalışmadığı zaman çalışır
        # Else bloğu - girintiyle oluşturulur.
        yapılacak işlemer
        
 Dikkat ederseniz burada else koşulunun yanına herhangi bir koşul yazmadık. Çünkü zaten else bloğunun çalışması için ondan önce gelen diğer koşulların sağlanmamasına bağlı.        

```python
yaş = int(input("Lütfen yaşınızı giriniz:"))
if (yaş < 18):
    print("Bu mekana giremezisinz.")
else:
    print("Bu mekana girebilirsiniz.")


                
                
```

# Koşullu Durumlar if - elif - else koşullu durumları

### if - elif - else Blokları

if koşulu :
        yapılacak işlemler
elif başka bir koşulu :
        yapılacak işlemler
elif başka bir koşul :
        yapılacak işlemler:
        //
        //
        //
else : 
    yapılacak işlemler

#### Programlarımızda, kaç tane koşulumuz var ise o kadar elif bloğu oluşturabiliriz. Ayrıca else'in yazılması zorunlu değildir. if - elif - else kalıplarında, hangi koşul sağlanırsa program o bloğu çalıştırır ve if - elif blokları sona erer. 

```python
### Kullanıcıya işlem seçtirdiğimiz program 
işlem = str(input("İşlem seçiniz:"))  # 3 tane işlem olsun
if işlem == "1" :
    print("1.işlem seçildi.")
elif işlem == "2" :
    print("İkinci işlem seçildi.")
elif işlem == "3" :
    print("üçüncü işlem seçildi.")
elif işlem == "4":
    print ("İşlem 4 seçildi.")
else:
    print("Geçersiz işlem.")
    
```

```python
#### BASİT HESAP MAKİNESİ :
print("""Basit hesap makinesi:)
işlemler;
1- toplama işlemi
2- çıkarma işlemi
3- bölme işlemi
4- çarpma işlemi""")
işlem = input("Lütfen işlemi giriniz:")
a = float(input("Lütfen bir sayı giriniz:"))
b = float(input("Lütfen bir sayı giriniz:"))
if işlem == "1":
    print("{} ve {}'nin Toplamı {}'dır".format(a,b,a+b))
elif işlem == "2":
    print("{} ve {}'nin çıkarılması {}'dır.".format(a,b,a-b))
elif işlem == "4":
    print("{} ve {}'nin çarpılması {}'dır".format(a,b,a * b))
elif işlem == "3":
    print("{} ve {}'nin bölünmesi {}'dır".format(a,b,a/b))
else:
    print("Geçersiz işlem....")

```

```python
print("""*********Kullanıcı girişi************** """)
sys_kullanıcı_adi = "Atakan"
sys_parola = "007"
kullanıcı_adi = input("Kullanıcı Adı:")
parola = input("Parola:")
if (kullanıcı_adi == sys_kullanıcı_adi and parola == sys_parola):
    print ("Hoşgeldiniz")
elif (kullanıcı_adi == sys_kullanıcı_adi and parola != sys_parola):
        print("Parola yanlış")
elif (kullanıcı_adi != sys_kullanıcı_adi and parola == sys_parola):
        print("Kullanıcı adı yanlış")
else:
    print("Kullanıcı adı ve parolanız yanlış.")
```

```python
####Beden Kitle İndeksi: Kilo / Boy(m) *  Boy(m)
 #BKİ 18.5'un altındaysa -------> Zayıf
 #BKİ 18.5 ile 25 arasındaysa ------> Normal
 #BKİ 25 ile 30 arasındaysa --------> Fazla Kilolu
 #BKİ 30'un üstündeyse -------------> Obez

boy = float(input("Boyunuzu metre cinsinden giriniz:"))
kilo = float(input("Kilonuzu giriniz:"))
bki = kilo/boy**2
if bki < 18.5:
    print("Zayıf")
elif 18.5 <= bki <= 25:
    print("Normal")
elif 25 <= bki <= 30:
    print("Fazla Kilolu")
elif bki >= 30:
    print("Obez")

```
