```python
#İkinci dereceden bir bilinmeyenli denklemin köklerini bulma
# Denklem ax² + bx + c
#Deltayı hesaplama = b**2 - 4*a*c
#birinci kök = (-b - delta ** 0.5) / (2*a)
#ikinci kök = (-b + delta ** 0.5) / (2*a)


a = int(input("a:"))
b = int(input("b:"))
c = int(input("c:"))
delta = b ** 2 - 4 * a * c
x1 = (-b - delta ** 0.5) / (2*a)
x2 = (-b + delta ** 0.5) / (2*a)
if delta < 0: print("delta cannot be negative")
elif delta == 0: print("delta equals 0")
elif delta > 0:
    print("Birinci kök : {}\nİkinci kök: {}\n".format(x1,x2))

```

```python
#Hipotenüsü bulmak :
#Hipotenüs Formülü: a^2 + b^2 = c^2

a = float(input("Lütfen bir sayı giriniz:"))
b = float(input("Lütfen bir sayı giriniz"))
c = a**2 + b**2
d = c ** 0.5
print("Birinci kenar uzunluğu: {},İkinci kenar uzunluğu: {},Hipotenüs: {}".format(a,b,d))
```
