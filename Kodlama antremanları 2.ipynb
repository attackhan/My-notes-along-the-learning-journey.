```python
print("""Kullanıcı giriş programına hoşgeldiniz Lütfen şifrenizi ve parolanızı doğru giriniz. 3 deneme hakkınız var""")
usernames = "Atakan"
passwords = "1234"
giris_hakkı = 3
while True:
    username = input("Lütfen Kullanıcı adını giriniz:")
    password = input("Lütfen parolanızı giriniz:")
    if username != usernames and password == passwords:
        print("Kullancı adı hatalı")
        giris_hakkı -= 1
    elif username == usernames and password != passwords:
        print("Parola hatalı")
        giris_hakkı -= 1
    elif username != usernames and password != passwords:
        print("Kullanıcı adı ve hatalı Parola hatalı")
        giris_hakkı -= 1
    else :
        print("Sisteme başarıyla giriş yapıldı.")
        break
    if  giris_hakkı == 0:
        print("Giriş hakkınız bitti")
        break
```

```python
#Kullanıcıdan aldığınız bir sayının mükemmel olup olmadığını bulmaya çalışın.
#Bir sayının kendi hariç bölenlerinin toplamı kendine eşitse bu sayıya "mükemmel sayı" denir.
# Örnek olarak, 6 mükemmel bir sayıdır. (1 + 2 + 3 = 6)

while True:
    a = input("Lütfen bir sayı giriniz çıkmak için 'q'ya basınız.")
    if a.lower() == "q":
        print("Programdan Çıkılıyor...")
        break
    elif not a.isdigit():
        print("Lütfen rakam giriniz")
        break
    sayı = int(a)
    toplam = 0

    i = 1
    while i < sayı:
        if sayı % i == 0:
            toplam += i
        i += 1
    if toplam == sayı:
            print("Sayı mükemmel sayıdır.")
    else:
            print("Sayı mükemmel sayı değildir.")
```

```python
#bir sayının "Armstrong" sayısı olup olmadığı

##Örnek olarak, Bir sayı eğer 4 basamaklı ise ve oluşturan rakamlardan herbirinin 4. kuvvetinin toplamı
#( 3 basamaklı sayılar için 3.kuvveti ) o sayıya eşitse bu sayıya "Armstrong" sayısı denir.##
##Örnek olarak : 1634 = 1^4 + 6^4 + 3^4 + 4^4

while True:
    a = input("Lütfen bir sayı giriniz(Çıkmak için'q'ya basınız):")
    if a.lower() == "q":
        print("Program'dan çıkış yapılıyor...")
        break
    elif not a.isdigit():
        print("Geçersiz bir değer girdiniz.")
        break
    else:
        sayı = int(a)
        basamak_sayisi = len(a)
        toplam = 0
        gecici_sayı = sayı

        ### Basamakları tek tek ayırmak için while döngüsü gerekir.
        while gecici_sayı > 0:
            basamak = gecici_sayı % 10 # Sayının en son basamağını verir.
            toplam += basamak ** basamak_sayisi #Basamağın n.kuvvetini toplama ekledik
            gecici_sayı //= 10 #Son basamağı sayıdan düşeriz (integer bölme)
        if toplam == sayı:
            print("Bu sayı 'Armstrong' sayısıdır. ")
        else:
            print("Bu sayı 'Armstrong' sayısı değildir.")


```

```python
##Her bir while döngüsünde kullanıcıdan bir sayı alın ve kullanıcının girdiği
# sayıları "toplam" isimli bir değişkene ekleyin.
## Kullanıcı "q" tuşuna bastığı zaman döngüyü sonlandırın ve
# ekrana "toplam değişkenini" bastırın.

print("""Toplam değişkeni. Çıkmak için 'q' ya basınız.""")
toplam = 0
while True:
    sayi = input("Lütfen bir sayı giriniz :")
    if sayi.lower() == "q":
        break
    sayi = int(sayi)
    toplam += sayi

print("Girdiğiniz sayıların toplamı :",toplam)
```
